```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve2d
from skimage import data, color, img_as_float

image = img_as_float(color.rgb2gray(data.astronaut()))

# Kernel 1
kernel1 = np.array([[1, 0, 1],
                    [0, -1, 0],
                    [1, 0, 1]]) 

# Kernel 2
kernel2 = np.array([[0, 0, 0],
                    [0, 200, 0],
                    [0, 0, 0]])  

# Kernel 3
kernel3 = np.array([[-1, -1, -1],
                    [-1, -1, -1],
                    [-1, -1, -1]])  

# Apply convolution to the image 
kernel1_image = convolve2d(image, kernel1, mode='same', boundary='symm')
kernel2_image = convolve2d(image, kernel2, mode='same', boundary='symm')
kernel3_image = convolve2d(image, kernel3, mode='same', boundary='symm')

# Plot the images
plt.figure(figsize=(12, 8))

plt.subplot(1, 4, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(kernel1_image, cmap='gray')
plt.title('Kernel 1')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(kernel2_image, cmap='gray')
plt.title('Kernel 2')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(kernel3_image, cmap='gray')
plt.title('Kernel 3')
plt.axis('off')

plt.tight_layout()
plt.show()
```

1. Give your observation, analysis and conclusion for each image.

Kernel 1 implements edge detection and highlights diagonal edges by comparing the center pixel with its diagonal neighbors. This results in a less saturated image, but more defined edges.

Kernel 2 implements amplification, multiplying each pixel by 200. This produces a brighter, highly-saturated image.

Kernel 3 implements an inversion by subtracting all neighboring pixel values from the center, producing a very dark or inverted-like image

2. Experiment modify what is the best kernel filter to enhance (sharpen) the image.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve2d
from skimage import data, color, img_as_float

image = img_as_float(color.rgb2gray(data.astronaut()))

# Kernel 4 (Sharpening)
kernel4 = np.array([[0, -1, 0],
                    [-1, 5, -1],
                    [0, -1, 0]])  

# Apply convolution to the image 
kernel4_image = convolve2d(image, kernel4, mode='same', boundary='symm')

# Plot the images
plt.figure(figsize=(15, 8))

plt.subplot(1, 1, 1)
plt.imshow(kernel4_image, cmap='gray')
plt.title('Kernel 4 (Sharpen)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
